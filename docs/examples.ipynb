# Some Examples

```python
import pyAERONET
```

Download AOD

```python
pyAERONET.download(station='Sao_Paulo',
         start_date='2010-01-01',
         end_date='2010-12-01',
         vars='AOD20',
         data_frequency='all')
```

Interpolate aod to 550nm

```python
import pandas as pd
```

```python
df = pd.read_csv('Sao_Paulo_AOD20.csv')
pyAERONET.compute_AOD550nm(df,return_columns=['time','AOD_550nm','440-870_Angstrom_Exponent']).head(10)
```

```python
pyAERONET.view_data(station='Sao_Paulo',
         start_date='2010-01-01',
         end_date='2014-01-01',
         vars='AOD20',
         data_frequency='all').head(10)
```

Compute broadband of single scattering albedo, asymmetry factor and surface albedo.

```python
# download the file of inversion
pyAERONET.download(station='Sao_Paulo',
         start_date='2010-01-01',
         end_date='2010-12-01',
         vars='ALL',
         inversion_type='ALM20',
         data_frequency='all')
```

```python
# open file and compute broadband
df_all = pd.read_csv('Sao_Paulo_ALL.csv')
pyAERONET.compute_broadband(df_all)[['time','ssa_broadband','asy_broadband','surface_albedo_broadband']].head(10)
```
